```python
#  we can also usr normal fuction

a = [1,2,3,4,5] #ittrable

def test(num):  # fuction
    return num*100
    
tuple(map(test,a)) 
```

```python
a = [1,2,3,4,5]

def test(num):
    if num%2==0:
        return 'even'
    else:
        return 'odd'
tuple(map(test,a))        
```

```python
# 2.filter(fuction,iterable)
# filter is conditon based
# it return only those for which condition result it true

nums= (10,-1,-5,4,-2,3)
list(filter(lambda n:n<0,nums))
```

```python
frd = ['pranav','nikhil','vinil','laxmi','anil']
tuple(filter(lambda nm:nm.endswith('il'),frd))
```
